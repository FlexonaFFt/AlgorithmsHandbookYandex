```python
def main():
    q_line = input().strip()
    while q_line == "":
        q_line = input().strip()
    q = int(q_line)

    lst = []
    out_lines = []

    for _ in range(q):
        line = input().strip()
        while line == "":
            line = input().strip()
        parts = line.split()
        t = int(parts[0])

        if t == 1:
            x = int(parts[1])
            y = int(parts[2])
            if x == 0:
                lst.insert(0, y)
            else:
                # вставка после x-го (1-based) → индекс x (0-based)
                lst.insert(x, y)

        elif t == 2:
            x = int(parts[1])
            out_lines.append(str(lst[x - 1]))

        elif t == 3:
            x = int(parts[1])
            lst.pop(x - 1)

    print("\n".join(out_lines))

if __name__ == "__main__":
    main()
```
