# O(n * n) solution

```python
class Solution:
    def finder(self, nums: list[int], x: int) -> int:
        for i, num in enumerate(nums):
            if num == x:
                return i + 1
        return -1

    def function(self):
        n, m = map(int, input().split())
        nums = list(map(int, input().split()))

        for _ in range(m):
            x = int(input().strip())
            print(self.finder(nums, x))

def main():
    solve = Solution()
    solve.function()

if __name__ == '__main__':
    main()
```

# O(n + q) solution

```python
import sys

def main():
    data = sys.stdin.read().strip().split()
    it = iter(data)

    # Читаем n и q
    n = int(next(it))
    q = int(next(it))

    # Читаем массив из n чисел
    nums = [int(next(it)) for _ in range(n)]

    # Предрасчёт: первая (самая левая) позиция каждого значения (1-based)
    first_pos = {}
    for i, num in enumerate(nums):
        # Сохраняем только первое вхождение
        if num not in first_pos:
            first_pos[num] = i + 1

    # Обрабатываем q запросов
    out_lines = []
    for _ in range(q):
        x = int(next(it))
        out_lines.append(str(first_pos.get(x, -1)))

    # Выводим все ответы разом — быстрее, чем много print()
    sys.stdout.write("\n".join(out_lines))

if __name__ == "__main__":
    main()
```
